## Machine Learning Classification Task

This notebook performs a baseline machine learning classification task using `scikit-learn` to predict 'User Behavior Class' from the provided dataset.

```python
# Import necessary libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.preprocessing import StandardScaler # Import StandardScaler
import joblib
```

```python
# 1. Load the dataset named 'user_behavior_dataset.csv'
df = pd.read_csv('/content/user_behavior_dataset.csv')

# Display the first few rows to understand the data
print("Original DataFrame head:")
display(df.head())
```

```python
# 2. Define target variable and drop 'User ID'
X = df.drop(columns=['User ID', 'User Behavior Class']) # Features
y = df['User Behavior Class'] # Target variable

print("\nFeatures (X) head after dropping User ID and target:")
display(X.head())
print("\nTarget (y) head:")
display(y.head())
```

```python
# 3. Perform data preprocessing: Apply One-Hot Encoding to categorical features
# Identify categorical columns for one-hot encoding
categorical_cols = ['Operating System', 'Gender', 'Device Model']

# Apply one-hot encoding using pd.get_dummies
X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True) # drop_first avoids multicollinearity

print("\nFeatures (X) head after One-Hot Encoding:")
display(X_encoded.head())
```

```python
# 4. Split the data into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
) # stratify=y ensures the same proportion of target class in train and test sets

# Scale numerical features after splitting to prevent data leakage
scaler = StandardScaler()
# Identify numerical columns in the encoded features (all non-boolean columns)
numerical_cols = X_train.select_dtypes(include=['number']).columns

X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])

print(f"\nTraining features shape: {X_train.shape}")
print(f"Testing features shape: {X_test.shape}")
print(f"Training target shape: {y_train.shape}")
print(f"Testing target shape: {y_test.shape}")
print("Numerical features scaled successfully.")
```

```python
# 5. Train a baseline Logistic Regression model
# Increase max_iter to ensure convergence, if a warning is thrown
model = LogisticRegression(max_iter=5000, random_state=42)
model.fit(X_train, y_train)

print("\nLogistic Regression model trained successfully.")
```

```python
# 6. Evaluate the model on the test set
y_pred = model.predict(X_test)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted') # 'weighted' handles potential class imbalance
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')

print(f"\nModel Evaluation on Test Set:")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
```

```python
# 7. Save the trained model into a file named 'baseline_model.pkl'
joblib.dump(model, 'baseline_model.pkl')

print("\nModel successfully saved as 'baseline_model.pkl'")
```
